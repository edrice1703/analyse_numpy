<b> Colonnes </b>

- Country: Nom du pays
- Density (P/Km2): Densité de population mesurée en personnes par kilomètre carré
- Agricultural Land (%): Pourcentage de la surface des terres utilisées à des fins agricoles
- Land Area (Km2): Superficie totale des terres du pays en kilomètres carrés
- Armed Forces Size: Taille des forces armées dans le pays
- Birth Rate: Nombre de naissances pour 1 000 habitants par an
- CO2 Emissions: Émissions de dioxyde de carbone en tonnes

# importation des données

```python
import numpy as np
data = np.genfromtxt('world-2023.csv',delimiter=',',dtype=None, encoding='utf-8-sig')
data
```

```python
entete = data[0,:]
pays = data[1:,:1]
donnee_numerique = data[1:,1:]
print(donnee_numerique)
```

```python
# supprimer le signe %
donnee_numerique[:,1] = np.char.replace(donnee_numerique[:,1], '%' , '').astype(float)

# changer le type des donnees en float
donnee_numerique = donnee_numerique.astype(float)
print(donnee_numerique)
```

```python
# selection de ligne avec nan
ligne_avec_nan = np.isnan(donnee_numerique).all(axis=1)

# pays avec valeur nan
pays_sans_donnees = pays[ligne_avec_nan]
print(pays_sans_donnees)
```

```python
# part nan sur colonne
p = np.isnan(donnee_numerique).sum(axis=0) / donnee_numerique.shape[0]
for i,entete in enumerate(entete[1:]):
    print(f"{entete} : {p[i]}")
```

```python
indice_pays_max_armee = np.nanargmax(donnee_numerique[:,2])
armee_la_plus_grande = np.nanmax(donnee_numerique[:,2])
pays_avec_plus_grande_armee = pays[indice_pays_max_armee]
print(pays_avec_plus_grande_armee)
print(armee_la_plus_grande)
```

```python
# moyenne et ecart_type du taux de natalité
moyenne = np.nanmean(donnee_numerique[:,3])
ecart_type = np.nanstd(donnee_numerique[:,3])
print(f"moyenne : {moyenne} , std : {ecart_type}")
```

```python
# les pays pollueurs (grande part de terre => grand pollueur ?)
part_terre_agricole = donnee_numerique[:,1]
emission_co2 = donnee_numerique[:,4]

moyenne_mondiale = np.nanmean(part_terre_agricole)
print(moyenne_mondiale)

# definitions de 25%
limite_emission_co2 = np.nanpercentile(emission_co2, 75)
id_haut_co2 = emission_co2 >= limite_emission_co2

print(limite_emission_co2)
```

```python
part_agricole_haut_co2 = part_terre_agricole[id_haut_co2]
comparaison_haut_co2 = np.nanmean(part_agricole_haut_co2) > moyenne_mondiale
print(comparaison_haut_co2)
print(np.nanmean(part_agricole_haut_co2))
print(moyenne_mondiale)
```
